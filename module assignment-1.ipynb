```python
X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).values

# Matrix-matrix multiplication
XTX = X.T @ X

# Invert XTX
XTX_inv = np.linalg.inv(XTX)

# Create y array
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Compute w
w = XTX_inv @ X.T @ y

# Sum of all elements in w
result_sum = w.sum()
result_sum
```
